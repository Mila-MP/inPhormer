# How to use the inPhormer python package. 
In this notebook, you'll find examples on how to use three different models to classify protein annotations based on their informativeness. There are three classes: 0 - uninformative, 1 - low informative, and 2- proper. For these examples we will be predicting the classes of three protein annotations; the uninformative "CSON008739 protein", the low informative "DUF499 domain-containing protein", and the proper "Phage portal protein lambda family". The three models used are a Tf-IDF vectorizer combined with a Complement Naive Bayes classifier, a pretrained embedder combined with a neural net classifier, and a fine tuned LLM. 

```python
# The models accepts lists, arrays, and slices of dataframes as an input
tests = ["CSON008s739 protein", "DUF499 domain-containing protein", "Phage portal protein lambda family"]
```

### TF-IDF + CNB

```python
from inphormer import TFIDFClassifier

# Load the pretrained TF-IDF + Complement Naive Bayes model
cnb = TFIDFClassifier.load_from_file()
```

```python
cnb.predict(tests)  # Predict the classes of the test annotations
```

### Embedder + NN

```python
from inphormer import EmbedderClassifier

# Load the pretrained Embedder + Neural Network model
embcls = EmbedderClassifier.load_model()
```

```python
embcls.predict(tests)  # Predict the classes of the test annotations
```

### Fine-tuned LLM

```python
from inphormer import BERTBasedModel

# Load the fine-tuned LLM model
model_link = "Mila-MP/inPhormer-LLM-classifier"
subfolder = "final_llm"
llm = BERTBasedModel(model_dir=model_link, subfolder=subfolder)
```

```python
llm.predict(tests)  # Predict the classes of the test annotations
```
